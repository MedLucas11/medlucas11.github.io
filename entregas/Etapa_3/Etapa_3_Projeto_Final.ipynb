## <u>Modelagem Funcional Geral</u>

O sistema desenvolvido consiste em uma fechadura eletrônica inteligente voltada ao controle de acesso em tempo real, baseada em Visão Computacional e Deep Learning. A literatura mostra que o reconhecimento facial é amplamente empregado em aplicações de segurança, vigilância e controle de acesso, inclusive em ambientes IoT e sistemas embarcados (SANCHEZ-MORENO et al., 2021; HORNG et al., 2022; ABDULLAHU; WACHE; PIANGERELLI, 2025; RAHIM et al., 2023).

Em termos funcionais, sistemas de reconhecimento facial normalmente se organizam em duas grandes tarefas: detecção facial e reconhecimento facial. Nos últimos anos, arquiteturas profundas, especialmente Redes Neurais Convolucionais (CNNs), tornaram-se predominantes nesse domínio por sua capacidade de aprender representações robustas para imagens faciais e melhorar a acurácia em cenários complexos (SANCHEZ-MORENO et al., 2021; EL FADEL, 2025; RAHIM et al., 2023).

O funcionamento do protótipo pode ser descrito em três etapas sequenciais. A primeira corresponde à aquisição e ao pré-processamento das imagens, em que a câmera captura continuamente os quadros da cena e o sistema realiza ajustes necessários para melhorar a qualidade da entrada visual. Essa etapa é relevante porque a literatura mostra que fatores como iluminação, pose, oclusão, escala e qualidade da captura afetam diretamente o desempenho do reconhecimento facial em ambientes não controlados (SANCHEZ-MORENO et al., 2021; AHSAN et al., 2021; ABDULLAHU; WACHE; PIANGERELLI, 2025).

A segunda etapa consiste na detecção e classificação facial. Nessa fase, a CNN localiza o rosto, extrai características discriminantes e compara essas informações com uma base de usuários cadastrados. Estudos recentes mostram o uso combinado de detectores faciais em tempo real e extratores profundos de características, como FaceNet, VGG16 e MTCNN, frequentemente associados a classificadores como SVM para viabilizar reconhecimento em tempo real com boa precisão (SANCHEZ-MORENO et al., 2021; ABDULLAHU; WACHE; PIANGERELLI, 2025; RAHIM et al., 2023).

A terceira etapa é a tomada de decisão. Nela, o sistema analisa o grau de compatibilidade entre a amostra capturada e o padrão armazenado e, caso o resultado supere um limiar de confiança previamente definido, autoriza o destravamento da fechadura. A adoção de limiares de decisão e mecanismos de aceitação, rejeição ou verificação adicional é compatível com modelos de autenticação facial baseados em threshold e decisões multicritério descritos na literatura (SHAH et al., 2023).

Para garantir a robustez do fluxo completo, a modelagem funcional deve estar apoiada em métricas quantitativas de avaliação. Trabalhos da área usam de forma recorrente medidas como acurácia, precisão, revocação, F1-score, AUC-ROC e desempenho temporal para verificar confiabilidade, segurança e viabilidade operacional de sistemas de visão computacional em tempo real (RAHIM et al., 2023; BABU et al., 2022; TRABELSI et al., 2023; AHSAN et al., 2021).

## <u>Método de Calibração</u>

Para preservar a integridade geométrica das imagens submetidas ao modelo de detecção facial, o sistema deve implementar uma etapa prévia de calibração de câmera e correção de distorções ópticas. A calibração é necessária para determinar a relação entre pontos tridimensionais do mundo e suas projeções bidimensionais na imagem, além de estimar parâmetros intrínsecos, extrínsecos e de distorção da lente, aspectos essenciais em aplicações de visão computacional (GUTIÉRREZ-MOIZANT et al., 2023).

O método adotado neste projeto é a Técnica de Calibração Baseada em Padrão Planar, conhecida como método de Zhang. Esse método é amplamente utilizado porque exige apenas a observação de um padrão plano em diferentes orientações, sem necessidade de aparatos tridimensionais complexos, sendo reconhecido pela literatura como uma solução prática, flexível e consolidada para calibração de câmeras (ZHANG, 2000; GUTIÉRREZ-MOIZANT et al., 2023; YONGJUN; ZUXUN; JIANQING, 2003; YONG-JUN; ZU-XUN; JIAN-QING, 2003).

#### 1. Justificativa da Escolha do Método
O método proposto por Zhang foi escolhido por sua viabilidade prática e robustez geométrica. Diferentemente de métodos clássicos que exigem estruturas de calibração mais rígidas e dispendiosas, essa técnica requer apenas que a câmera observe um padrão planar, como um tabuleiro de xadrez, em algumas orientações diferentes, podendo tanto a câmera quanto o padrão serem movimentados livremente durante a coleta (ZHANG, 2000; YONGJUN; ZUXUN; JIANQING, 2003).

Além disso, o método modela explicitamente a distorção radial da lente e combina uma solução inicial em forma fechada com refinamento não linear, o que contribui para resultados confiáveis em dados simulados e reais (ZHANG, 2000). Trabalhos posteriores continuam usando o método de Zhang como base para melhorias em contextos específicos, como câmeras desfocadas e estratégias otimizadas de aquisição de imagens de calibração, o que reforça seu papel como referência metodológica na área (HUO et al., 2022; HAO; TAI; TAN, 2023).

#### 2. Fundamentação Teórica da Calibração
O processo de calibração busca relacionar as coordenadas do mundo tridimensional com o plano bidimensional da imagem digital, baseando-se no modelo matemático da Câmera Escura (*Pinhole Camera Model*). O objetivo central é isolar dois conjuntos de parâmetros:

* **Matriz Intrínseca ($K$):** Modela a geometria interna da câmera. É responsável por projetar os raios de luz no sensor digital, abrigando informações vitais como a distância focal ($f_x, f_y$) e as coordenadas exatas do centro óptico da imagem ($c_x, c_y$).
* **Coeficientes de Distorção ($D$):** Lentes comerciais grandes-angulares (comuns em webcams) não são perfeitas e introduzem aberrações na captura da luz. A distorção radial curva as linhas retas nas bordas da imagem (conhecido como efeito barril ou almofada), enquanto a distorção tangencial ocorre quando a lente não está perfeitamente paralela ao sensor. Os coeficientes ($k_1, k_2, k_3, p_1, p_2$) criam um modelo polinomial capaz de reverter esses desvios.

Para uma aplicação de **Detecção Facial**, a mitigação dessas distorções é um pré-requisito crítico. Um rosto localizado nas extremidades de uma imagem não calibrada sofrerá alongamentos ou compressões artificiais. Ao retificar a imagem matematicamente, o sistema entrega à CNN um rosto com proporções anatômicas reais, independentemente de sua posição na tela, otimizando a extração de características e melhorando a acurácia global do modelo.

#### 3. Procedimento Metodológico
A execução do processo de calibração e correção contínua seguirá o seguinte fluxo algorítmico:

1. **Aquisição de Dados Espaciais:** Captura prévia de múltiplas amostras visuais (entre 15 e 20 imagens) do padrão de calibração em variadas posições e ângulos de inclinação, assegurando que o padrão cubra todas as extremidades do campo de visão da lente.
2. **Extração de Características Bidimensionais:** Aplicação de algoritmos de visão computacional para identificar as coordenadas de pixel exatas correspondentes às interseções do tabuleiro de xadrez nas imagens capturadas.
3. **Otimização Não-Linear:** Estabelecimento da relação geométrica entre os pontos 3D ideais do padrão e os pontos 2D extraídos. Utiliza-se um algoritmo de minimização matemática para reduzir o erro de reprojeção global, calculando assim a Matriz Intrínseca e os Coeficientes de Distorção definitivos.
4. **Retificação de Imagem (Pré-processamento da CNN):** Durante a execução ao vivo do sistema de detecção facial, os parâmetros extraídos são utilizados para gerar mapas de remapeamento. Cada frame de vídeo capturado sofrerá uma transformação geométrica corretiva em tempo real milissegundos antes de ser processado pela Rede Neural Convolucional.



## <u>Método de Avalição Funcional</u>



Para a validação do sistema, é necessário adotar métricas objetivas que permitam avaliar sua robustez, sua confiabilidade e sua capacidade de operar em tempo real. A literatura da área de reconhecimento facial e visão computacional usa recorrentemente métricas como precisão, revocação, F1-score, acurácia, AUC-ROC e medidas temporais para comparar modelos e verificar desempenho experimental(RAHIM et al., 2023; BABU et al., 2022; TRABELSI et al., 2023; AHSAN et al., 2021).

A primeira métrica considerada é a latência de processamento, medida em quadros por segundo (**FPS**), que reflete o desempenho temporal do sistema. O FPS indica a capacidade de operar de forma fluida em tempo real durante toda a sequência de captura da imagem, correção de distorção, detecção facial e decisão de acesso, evitando atrasos que comprometeriam a experiência de uso. Essa métrica é calculada pela razão entre o número total de quadros processados e o tempo total gasto nesse processamento, conforme a expressão:

$$FPS = \frac{N_{quadros}}{T_{total}}$$

onde $N_{quadros}$ representa o número de quadros processados e $T_{total}$ o tempo total decorrido em segundos. Esta medição será realizada em diferentes condições de iluminação para avaliar o impacto do ambiente sobre a eficiência do processamento.

A avaliação da qualidade do reconhecimento facial, por sua vez, baseia-se na construção de uma matriz de confusão para a tarefa de classificação entre acesso autorizado e não autorizado. Cada tentativa de destravamento realizada por voluntários será classificada em uma de quatro categorias:

1. **Verdadeiro Positivo (VP):** quando um usuário autorizado é corretamente reconhecido e tem seu acesso liberado;
2. **Falso Positivo (FP):** quando um usuário não autorizado é indevidamente reconhecido como autorizado (uma falha crítica de segurança);
3. **Falso Negativo (FN):** quando um usuário autorizado não é reconhecido e tem seu acesso negado indevidamente (uma falha de usabilidade);
4. **Verdadeiro Negativo (VN):** quando um usuário não autorizado é corretamente rejeitado pelo sistema.

A partir dessas contagens, determinam-se a **Precisão** — proporção de acessos liberados que de fato pertencem a usuários legítimos — e a **Revocação** — proporção de usuários autorizados que foram corretamente identificados. Ambas são calculadas pelas seguintes fórmulas:

$$Precisão = \frac{VP}{VP + FP}$$

$$Revocação = \frac{VP}{VP + FN}$$

Como métrica central de qualidade do reconhecimento, destaca-se o **F1-Score**, definido como a média harmônica entre a precisão e a revocação:

$$F1\text{-}Score = 2 \times \frac{Precisão \times Revocação}{Precisão + Revocação}$$

O F1-Score resulta em um valor único entre 0 e 1 que penaliza de forma equilibrada tanto os falsos positivos quanto os falsos negativos, oferecendo uma visão consolidada da confiabilidade do sistema.

Por fim, será calculado o erro médio de localização facial, expresso em pixels. Essa métrica avalia a precisão espacial da detecção ao comparar a posição do centro do rosto detectado pelo algoritmo com uma posição de referência anotada manualmente em um conjunto de quadros de teste.

Para validar experimentalmente todas essas métricas, será constituído um conjunto de teste composto por voluntários autorizados e não autorizados, com múltiplas capturas por pessoa em variadas condições de ângulo, distância e iluminação. O sistema será executado em tempo real sobre essa base de dados, registrando automaticamente os resultados de cada tentativa para consolidar, ao final, o FPS médio, o erro médio de localização e os valores da matriz de confusão com suas métricas derivadas.
