### **Resumo das Métricas de Avaliação Final**

Com base na pesquisa aplicada após os testes práticos, foram coletadas 9 respostas baseadas em uma escala de avaliação de 1 a 5. Os resultados quantitativos atestam a excelente viabilidade e aceitação do sistema:

**Aceitação e Facilidade de Uso**
* 100% dos voluntários (9 de 9) atribuíram nota máxima (5) ao afirmarem que gostariam de usar o sistema com frequência.
* 100% dos voluntários (9 de 9) avaliaram com nota máxima (5) a facilidade de uso do sistema.
* 100% dos voluntários (9 de 9) sentiram-se confiantes ao operar o sistema, atribuindo nota máxima (5).
* 100% dos voluntários (9 de 9) deram nota máxima (5) para a interatividade do sistema.
* 100% dos voluntários (9 de 9) concordaram totalmente (nota 5) que a maioria das pessoas aprenderia a usar o sistema rapidamente.
* 100% dos voluntários (9 de 9) avaliaram com nota máxima (5) que as funções do sistema foram bem integradas.

**Complexidade e Curva de Aprendizado**
* A totalidade da amostra (9 de 9) discordou totalmente, atribuindo a menor nota possível (1), da afirmação de que o sistema é desnecessariamente complexo.
* A totalidade da amostra (9 de 9) avaliou com a nota mínima (1) a afirmação de que o sistema é complicado de usar.
* Todos os participantes (9 de 9) atribuíram nota mínima (1) quanto à percepção de inconsistências no sistema.
* A independência do usuário foi confirmada: 8 testadores deram nota mínima (1) e apenas 1 deu nota 2 ao serem questionados se precisariam de suporte técnico para usar o sistema.
* A curva de aprendizado foi mínima: 8 testadores atribuíram nota mínima (1) e apenas 1 deu nota 2 ao afirmarem que precisaram aprender muitas coisas antes de poder usar a aplicação.

### **Avaliação de Usabilidade e Testes com Voluntários**


#### 1. Usabilidade e Curva de Aprendizado
Os resultados quantitativos demonstraram um alto índice de aceitação e facilidade de operação:
* Todos os 9 voluntários avaliaram com nota máxima (5/5) a facilidade de uso do sistema.
* A totalidade dos participantes também atribuiu nota máxima (5/5) para a sensação de confiança ao utilizar a aplicação e para a interatividade do sistema.
* Em contrapartida, todos os 9 testadores atribuíram a nota mínima (1/5) ao serem questionados se acharam o sistema complicado de usar, desnecessariamente complexo ou se possuía inconsistências.
* A curva de aprendizado provou-se baixa, com 8 dos 9 voluntários indicando que não precisariam do suporte de uma pessoa técnica para operar o sistema. Além disso, 8 dos 9 usuários discordaram totalmente (nota 1/5) da afirmação de que precisaram aprender muitas coisas antes de usar a ferramenta.

#### 2. Destaques Qualitativos: Interface e UX
A interface gráfica foi o ponto de maior destaque positivo apontado pelos testadores.
* Quando questionados sobre o que mais gostaram no sistema, 8 das 9 respostas citaram diretamente a "interface" ou "UI".
* Os voluntários utilizaram adjetivos como "amigável" e "clear", e também elogiaram a "facilidade de uso", a "velocidade e indicadores" apresentados no frontend.

#### 3. Oportunidades de Melhoria e Feedbacks
As perguntas abertas também capturaram pontos de atenção importantes para a evolução do projeto:
* **Captação de Imagem:** Um testador relatou "uma leve dificuldade inicial na captura do rosto", e outro mencionou que o que menos gostou foi ter que "aproximar muito o rosto" da câmera.
* **Segurança e Implantação:** Na seção de sugestões, foram recomendadas ações como "verificar questões de segurança (vulnerabilidades)" e "hospedar o site em algum domínio", visando uma aplicação comercial em ambiente web.

#### 4. Validação Prática de Segurança
Os testes realizados pelos voluntários comprovaram o pleno funcionamento da arquitetura de segurança do sistema:
* Os usuários compreenderam perfeitamente o escopo do projeto, definindo-o corretamente como uma ferramenta para "identificação e segurança", "validar rostos e liberar acessos" e agir como uma "fechadura digital".
* Os voluntários descreveram experimentos práticos que incluíram o cadastro no sistema, testes com reconhecimento facial e avaliação dos logs de acesso.
* Um dos cenários de estresse fundamentais foi validado na prática por um usuário que relatou: "Eu tentei reconhecer a minha face sem registrá-la e o sistema me recusou, depois, com a face cadastrada, reconheceu".
* Os resultados obtidos nos experimentos foram descritos como bem-sucedidos por todos os testadores, gerando a conclusão de que o sistema "recusou quando devia e liberou quando esperado".
