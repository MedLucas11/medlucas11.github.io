## 1. Avaliação de Desempenho Matemático (Testes Offline)

Para validar a eficácia e a segurança da inteligência artificial aplicada ao sistema, foi realizado um teste em lote (*offline*) com um total de 955 imagens. O objetivo principal desta etapa foi encontrar o equilíbrio ótimo entre a segurança patrimonial (barrar intrusos) e a conveniência do usuário (liberar moradores).

O sistema identificou automaticamente que o **Limiar (*Threshold*) Ideal** para a aplicação é de **0.85** (distância euclidiana miníma para dar um "match" com o embedding). Operando sob este parâmetro de rigor, o modelo obteve os seguintes resultados nas predições:

*   **Verdadeiros Positivos (Acesso Correto):** 626 validações.
*   **Verdadeiros Negativos (Bloqueio Correto):** 291 validações.
*   **Falsos Positivos (Risco de Segurança):** Apenas 9 ocorrências.
*   **Falsos Negativos (Risco de Experiência):** 29 ocorrências.

A partir desta matriz de confusão, o sistema atingiu métricas sólidas de desempenho computacional:
*   **Acurácia (96.0%):** O modelo acertou a grande maioria das classificações gerais.
*   **Precisão (98.6%):** Sendo a métrica mais crítica para uma fechadura, este valor quase perfeito demonstra que, quando o sistema afirma que alguém tem acesso liberado, ele está quase sempre correto, minimizando drasticamente o risco de invasões.
*   **Recall (95.6%):** Indica uma excelente capacidade de reconhecer os verdadeiros donos da casa, com uma taxa de rejeição falsa muito baixa.
*   **F1-Score (97.1%):** A média harmônica comprova que o modelo encontrou um estado da arte no equilíbrio entre a Precisão e o Recall. O sacrifício de gerar 29 Falsos Negativos (obrigando o morador a tentar a leitura uma segunda vez) é perfeitamente justificável para garantir apenas 9 Falsos Positivos em quase mil tentativas.

## 2. Avaliação de Usabilidade e Testes com Voluntários

A viabilidade prática do projeto foi aferida através de uma pesquisa aplicada a 9 voluntários após a utilização prática do sistema. O questionário utilizou uma escala de avaliação de 1 a 5 e obteve resultados que atestam a excelente aceitação da arquitetura.

### 2.1. Curva de Aprendizado e Aceitação
*   A facilidade de uso foi avaliada com nota máxima (5) por 100% dos voluntários (9 de 9).
*   A sensação de confiança ao operar o sistema também recebeu nota máxima (5) da totalidade dos participantes.
*   A curva de aprendizado provou-se baixa, com 8 dos 9 voluntários indicando que não precisariam do suporte de uma pessoa técnica para operar o sistema.
*   A afirmação de que precisaram aprender muitas coisas antes de usar a ferramenta foi totalmente discordada (nota 1) por 8 dos 9 usuários.
*   Todos os 9 testadores atribuíram a nota mínima (1) ao serem questionados se acharam o sistema complicado de usar, desnecessariamente complexo ou se possuía inconsistências.

### 2.2. Destaques Qualitativos: Interface e UX
*   A interface gráfica foi o ponto de maior destaque positivo apontado pelos testadores.
*   Ao relatarem o que mais gostaram no sistema, 8 das 9 respostas citaram diretamente a "interface" ou "UI".
*   Os voluntários utilizaram adjetivos como "amigável" e "clear", elogiando também a "facilidade de uso", a "velocidade e indicadores" apresentados no frontend.

### 2.3. Oportunidades de Melhoria
*   **Captação de Imagem:** Um testador relatou "uma leve dificuldade inicial na captura do rosto", e outro mencionou que o que menos gostou foi ter que "aproximar muito o rosto" da câmera.
*   **Segurança e Implantação:** Foram recomendadas ações como "verificar questões de segurança (vulnerabilidades)" e "hospedar o site em algum domínio", visando uma aplicação comercial em ambiente web.

### 2.4. Validação Prática de Segurança
*   Os voluntários descreveram os experimentos práticos englobando o cadastro no sistema, testes com reconhecimento facial e avaliação dos logs de acesso.
*   Os usuários definiram o escopo corretamente como uma ferramenta para "identificação e segurança", "validar rostos e liberar acessos" e agir como uma "fechadura digital".
*   Um cenário de estresse fundamental foi validado empiricamente, conforme relatado por um usuário: "Eu tentei reconhecer a minha face sem registrá-la e o sistema me recusou, depois, com a face cadastrada, reconheceu".
*   A conclusão geral dos testadores foi de que os experimentos foram bem-sucedidos e o sistema "recusou quando devia e liberou quando esperado".

## 3. Conclusão Final

A triangulação dos resultados obtidos nos testes matemáticos (offline) e nos testes de campo com usuários (online) comprova o sucesso no desenvolvimento do Sistema de Fechadura Eletrônica com Detecção Facial (MVP). O projeto atendeu de forma plena ao seu objetivo de criar uma fechadura digital robusta e funcional baseada em Visão Computacional.

Do ponto de vista sistêmico, as redes neurais e a lógica de distância espacial alcançaram alto grau de maturidade estatística. Com um limiar configurado em 0.85, a precisão matemática de 98.6% garantiu uma arquitetura altamente restritiva contra invasores (apenas 0,94% de erro em 955 testes), sem inviabilizar a usabilidade.

Na perspectiva do usuário final (UX/UI), a alta precisão matemática refletiu-se em uma operação fluida. A integração do backend analítico com um frontend intuitivo e reativo via WebSockets resultou em índices absolutos (100% de aceitação) nos quesitos de confiança, facilidade e interatividade. O sistema converteu com maestria a complexidade das extrações vetoriais em indicadores visuais simples e de rápida interpretação para o cidadão comum.

Em suma, a aplicação é altamente segura como exige uma fechadura de controle de acesso, e amigável o suficiente para ser adotada no dia a dia, validando completamente a viabilidade técnica e prática do produto proposto.

### 3.1. Oportunidades de Melhoria e Trabalhos Futuros

Apesar dos excelentes resultados, a validação deste MVP permitiu mapear oportunidades de evolução técnica para transformar o protótipo em uma solução comercial escalável:

*   **Prevenção de Fraudes (*Liveness Detection*):** A principal evolução de segurança seria a implementação de algoritmos anti-spoofing para verificar a "vivacidade" do usuário, impedindo que o sistema seja enganado por fotos impressas ou vídeos reproduzidos em telas de celulares.
*   **Arquitetura e Hospedagem (IoT):** Migrar a aplicação de um ambiente local para um modelo distribuído, hospedando o banco de dados e a interface em nuvem, enquanto a inferência pode ser embarcada em dispositivos de borda (*Edge Computing*), como um Raspberry Pi.
*   **Melhorias de Hardware:** A utilização de câmeras com lentes de menor distorção ou sensores de profundidade e infravermelho resolveria os atritos relatados na captação, evitando que o usuário precise se aproximar demasiadamente da lente e permitindo o uso eficiente em ambientes com baixa luminosidade.
*   **Auditoria de Cibersegurança:** Realizar auditorias aprofundadas e testes de intrusão na comunicação via WebSockets e na criptografia do armazenamento do banco de dados biométrico (*embeddings*), garantindo a integridade dos dados sensíveis dos moradores.

Em suma, a aplicação atual é altamente segura como exige uma fechadura de controle de acesso e amigável o suficiente para ser adotada no dia a dia, consolidando uma base técnica sólida para futuras expansões e integrações em ambientes de casas inteligentes.
