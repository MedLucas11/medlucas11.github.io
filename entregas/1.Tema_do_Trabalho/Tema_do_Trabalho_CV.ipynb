## <u>Descrição do Contexto</u>

O Reconhecimento Facial (FR) é uma tecnologia essencial utilizada para identificar e verificar indivíduos com base em suas características faciais. Nas últimas décadas, o FR assumiu um papel crucial em diversos setores, incluindo segurança, saúde, sistemas bancários e identificação criminal, com o desenvolvimento contínuo de técnicas que variam de abordagens focadas em características locais a métodos holísticos e híbridos (Gururaj, H. L., et al., 2024). Paralelamente, o interesse na formulação de teorias e algoritmos para sistemas de reconhecimento tem crescido rapidamente, impulsionado por aplicações concretas na indústria, como o controle de acesso a edifícios e o videomonitoramento (Kortli, Y., et al., 2020). Inserido nesse cenário tecnológico atual, o presente trabalho define como tema o desenvolvimento de um sistema de reconhecimento facial aplicado a uma fechadura eletrônica, visando modernizar o controle de acesso por meio da biometria passiva.

O avanço expressivo na precisão e na eficiência dessa tecnologia deve-se, em grande parte, à evolução do deep learning. A capacidade das Redes Neurais Convolucionais (CNNs) de aprender hierarquias espaciais e de reconhecer rostos sob diferentes posturas e situações de iluminação popularizou amplamente o seu uso, com modelos atingindo o estado da arte em benchmarks desafiadores (Farfade et al., 2015; Minaee et al., 2021). Contudo, a literatura aponta que a aplicação desses modelos em contextos do mundo real (in-the-wild) ainda enfrenta obstáculos. Reconhecer faces e expressões apresenta uma queda de generalização devido à diversidade humana, como variações de idade e etnia, além de drásticas diferenças nas condições de captura, como iluminação do ambiente e oclusões causadas por acessórios ou ângulos de câmera (Ramis, S., et al., 2022). Em ambientes de segurança predial ou laboratorial, os algoritmos estáticos tradicionais falham frequentemente diante de capturas de baixa resolução, o que exige a adoção de frameworks de redes neurais mais robustos para a identificação correta dos sujeitos (Ren, Z., et al., 2025).

Além dos desafios visuais, a implementação deste projeto em uma fechadura eletrônica esbarra em restrições físicas de hardware. Embora haja grande progresso na detecção facial precisa, o alto custo computacional e o peso dos modelos baseados em deep learning dificultam sua implantação em dispositivos de borda (edge devices) e sistemas embarcados móveis, onde a capacidade de memória e a latência são altamente restritas. Para que o sistema seja viável como um produto, torna-se imperativa a utilização de detectores de face otimizados que operem na escala de milissegundos (Wu, W., et al., 2023), garantindo o processamento de imagens em tempo real para a rápida liberação da porta.

#### Originalidade
A originalidade do trabalho consiste em integrar métodos modernos de detecção de rostos com as rigorosas restrições de processamento de um dispositivo físico na borda (edge device), como uma fechadura eletrônica. Modelos pesados e com alto custo computacional dificultam a implantação em dispositivos móveis e embarcados, onde o tamanho do modelo e a latência são altamente restritos.

## <u>Tema Escolhido e Justificativa</u>



Diante das entrevistas, o tema escolhido foi:

**Sistema de Deteccção Facial em Fechaduras Eletrônicas: Mitigação de Falhas e Redução de Atrito.**

A definição deste projeto fundamenta-se diretamente nas dores e receios levantados durante a Fase de Empatia. Ao analisar as entrevistas com os potenciais usuários, identificou-se um contraste expressivo: embora exista um claro atrito na rotina diária com chaves físicas — especialmente quando os moradores chegam em casa com as mãos ocupadas por sacolas, lixo ou animais de estimação —, há uma forte resistência à adoção da biometria facial devido à falta de confiabilidade percebida nos sistemas eletrônicos atuais.

A dor comum que unifica os discursos dos entrevistados é o medo da falha técnica e as suas consequências diretas na segurança pessoal. Relatos apontaram que experiências prévias ou a simples percepção de processos eletrônicos falhos geram o receio de que o morador fique trancado para fora ou, mais gravemente, passe muito tempo exposto na rua, vulnerável a abordagens criminosas enquanto o sistema tenta validar o acesso. Além disso, destacou-se a preocupação com momentos de inoperabilidade (como quedas de energia) e a necessidade de que a tecnologia não crie barreiras para idosos ou pessoas com deficiência.

Dessa forma, a justificativa e a motivação primária deste trabalho consistem em atacar diretamente essa desconfiança. O projeto não busca apenas substituir uma chave por uma câmera, mas sim desenvolver um sistema de visão computacional focado em alta confiabilidade e contingência. A motivação é provar que, ao projetar um modelo robusto que neutralize os falsos negativos e opere com rapidez, é possível converter o ceticismo e o sentimento de vulnerabilidade do usuário em conveniência e segurança, viabilizando a adoção da tecnologia sem o risco de exclusão digital ou abandono em momentos críticos.

## <u>Cenário de Aplicação e Benefícios</u>


A evolução das tecnologias de segurança e a busca por soluções mais eficientes e convenientes têm impulsionado a modernização do controle de acesso a ambientes restritos. Nesse cenário, o desenvolvimento de uma fechadura eletrônica baseada em visão computacional e detecção facial surge como uma alternativa inovadora aos métodos tradicionais, encontrando aplicabilidade em diversos setores. Em ambientes residenciais (smart homes), o sistema substitui chaves físicas e tags, proporcionando autonomia aos moradores. No ambiente corporativo, a tecnologia viabiliza o controle rigoroso de fluxo em salas de servidores, laboratórios e escritórios que manipulam dados sensíveis, eliminando o risco de compartilhamento indevido de crachás. Além disso, o sistema se mostra ideal para a gestão de áreas comuns em condomínios, onde o fluxo intenso de pessoas exige um cadastro dinâmico, rápido e seguro.

A integração tecnológica proposta trará melhorias práticas para a segurança e a gestão de acessos. Ao substituir senhas e cartões RFID, suscetíveis a perdas ou clonagens, por características faciais únicas, o sistema utilizará algoritmos de deep learning para garantir que apenas indivíduos autorizados entrem no local. Por ser uma tecnologia sem contato (touchless), oferecerá conveniência e higiene, permitindo a liberação rápida da porta mesmo quando o usuário estiver com as mãos ocupadas. A gestão será centralizada e digital: o cadastro e a revogação de permissões ocorrerão diretamente na base de dados, eliminando a substituição de componentes físicos, enquanto todas as tentativas de acesso serão registradas com data, hora e identificação para viabilizar futuras auditorias.

Para sustentar essa operação, a infraestrutura técnica do projeto garantirá um processamento de baixa latência por meio de modelos computacionais eficientes, que acionarão a fechadura na ordem de milissegundos e manterão uma taxa de quadros (FPS) fluida para interações em tempo real. Além disso, a solução assegurará robustez em condições variáveis, aplicando processos de calibração para corrigir distorções de lente da câmera, aliados ao uso de redes neurais capazes de identificar os usuários com precisão sob diferentes iluminações e ângulos de posicionamento facial.
